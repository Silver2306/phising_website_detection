```python
import joblib
import pandas as pd
```

```python
model = joblib.load("phissafe_random_forest.joblib")
features = joblib.load("phissafe_features.joblib")

print("Model loaded successfully")
print("Expected features:", len(features))
print(features)
```

```python
df = pd.read_csv("../data/data.csv")

df = df.drop_duplicates(subset=["URL"])
```

```python
X = df[features]
y = df["label"]
```

Running a test on one row

```python
sample = X.iloc[[0]]

print(sample)
```

```python
prediction = model.predict(sample)

print("Prediction:", prediction[0])
print("Actual:", y.iloc[0])
```

```python
if prediction[0] == 0:
    print("Predicted: Phishing")
else:
    print("Predicted: Legitimate")

if y.iloc[0] == 0:
    print("Actual: Phishing")
else:
    print("Actual: Legitimate")
```

running tests on multiple rows

```python
samples = X.iloc[:10]

predictions = model.predict(samples)

for i in range(10):
    print(
        "Prediction:", predictions[i],
        "| Actual:", y.iloc[i]
    )
```
